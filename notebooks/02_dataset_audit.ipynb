```python
from pathlib import Path
import platform
import sys

PROJECT_ROOT = Path.cwd().parent
DATA_DIR = PROJECT_ROOT / "data"
RAW_DIR = DATA_DIR / "raw"
PROCESSED_DIR = DATA_DIR / "processed"

print("Python:", sys.version.split()[0])
print("Platform:", platform.platform())
print("Project root:", PROJECT_ROOT)
print("Raw data directory:", RAW_DIR)
print("Processed data directory:", PROCESSED_DIR)

print("\nRaw directory exists:", RAW_DIR.exists())
print("Processed directory exists:", PROCESSED_DIR.exists())
```

```python
# Create required data directories

RAW_DIR.mkdir(parents=True, exist_ok=True)
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print("Created:")
print("✓", RAW_DIR)
print("✓", PROCESSED_DIR)

print("\nVerification:")
print("Raw directory exists:", RAW_DIR.exists())
print("Processed directory exists:", PROCESSED_DIR.exists())
```

```python
# AEGIS dataset metadata schema

import pandas as pd

columns = [
    "sample_id",
    "file_path",
    "category",
    "visual_label",
    "audio_label",
    "overall_label",
    "duration_sec",
    "fps",
    "frame_count",
    "width",
    "height",
    "audio_sample_rate",
    "audio_channels",
    "source_subject",
    "split",
]

metadata = pd.DataFrame(columns=columns)

print("AEGIS metadata schema")
print("=" * 60)
print("Number of fields:", len(metadata.columns))
print("\nFields:")
for i, column in enumerate(metadata.columns, start=1):
    print(f"{i:02d}. {column}")
```

```python
# Save the metadata schema for future dataset ingestion

METADATA_PATH = PROCESSED_DIR / "aegis_metadata_schema.csv"

metadata.to_csv(METADATA_PATH, index=False)

print("Schema saved to:")
print(METADATA_PATH)

print("\nFile exists:", METADATA_PATH.exists())
```

```python
# Verify that the saved schema can be loaded correctly

verified_metadata = pd.read_csv(METADATA_PATH)

print("Loaded successfully:", METADATA_PATH.exists())
print("Shape:", verified_metadata.shape)

print("\nColumns:")
print(verified_metadata.columns.tolist())
```
